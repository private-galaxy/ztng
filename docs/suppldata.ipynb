# Supplementary data catalogs

`ztng.suppldata` reads the post-processing catalogs that the IllustrisTNG data
release publishes on its website. It currently covers three of them:

- [(c) Stellar Circularities, Angular Momenta, Axis Ratios](https://www.tng-project.org/data/docs/specifications/#sec5c)
- [(g) Stellar Assembly](https://www.tng-project.org/data/docs/specifications/#sec5g)
- [(q) Halo Structure](https://www.tng-project.org/data/docs/specifications/#sec5q)

Each subsection gives the download link and describes every field with its
unit. Put the downloaded files in this layout, next to the `output/` directory
of the run:

```
~/data/simulations/TNG100-1/supplementary/
|- stellar_circularities/tng100-1_stellar_circs.hdf5
|- stellar_assembly/tng100-1_stellar_assembly.hdf5
'- halo_structure/tng100-1_halo_structure_099.hdf5
```

````python
import os

from ztng.suppldata.local import (
    getHaloStructure,
    getHaloesStructure,
    getSubhaloStellarAssembly,
    getSubhaloStellarCirc,
    getSubhaloesStellarAssembly,
    getSubhaloesStellarCirc,
)

root = os.path.expanduser('~/data/simulations')
catalogs = {
    '(c) stellar_circs': (
        f'{root}/TNG100-1/supplementary/stellar_circularities/'
        'tng100-1_stellar_circs.hdf5'),
    '(g) stellar_assembly': (
        f'{root}/TNG50-4/supplementary/stellar_assembly/'
        'tng50-4_stellar_assembly.hdf5'),
    '(q) halo_structure': (
        f'{root}/TNG100-1/supplementary/halo_structure/'
        'tng100-1_halo_structure_099.hdf5'),
}
for label, path in catalogs.items():
    status = 'ready' if os.path.exists(path) else f'missing: {path}'
    print(f'{label:22s}', status)
````

## (c) Stellar Circularities, Angular Momenta, Axis Ratios

````python
# Read the stellar circularities of all subhaloes in one snapshot.
getSubhaloesStellarCirc('TNG100-1', 99)
````

`getSubhaloStellarCirc` takes a Subfind ID and reads only that row.

````python
# Read the stellar circularities of one subhalo in one snapshot.
getSubhaloStellarCirc('TNG100-1', 99, 266560)
````

## (g) Stellar Assembly

````python
# Read the stellar assembly of all subhaloes in one snapshot.
getSubhaloesStellarAssembly('TNG50-4', 99)
````

````python
# Read the stellar assembly of one subhalo in one snapshot.
getSubhaloStellarAssembly('TNG50-4', 99, 7)
````

## (q) Halo Structure

One row per FoF halo, so a row number is a group number. Most haloes are below
the mass threshold of the catalog, they carry `GroupFlag == 0` and NaN
properties, and the single halo reader warns about it:

````python
# Read the structural properties of all FoF haloes in one snapshot.
haloes = getHaloesStructure('TNG100-1', 99)
print(len(haloes['GroupFlag']), int((haloes['GroupFlag'] == 1).sum()))
del haloes
````

````python
import warnings

# Read the structural properties of one FoF halo in one snapshot.
getHaloStructure('TNG100-1', 99, 500)

# A halo below the mass threshold warns and returns NaN.
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter('always')
    below = getHaloStructure('TNG100-1', 99, 6291348)

print(caught[0].message)
int(below['GroupFlag']), below['M200c']
````

## When something is missing

A catalog that is not on disk raises `FileNotFoundError`, and the message
gives the path it expected and the section of the specifications. An object
that the catalog does not list raises `KeyError`.

````python
try:
    getHaloesStructure('TNG300-1', 99)
except FileNotFoundError as error:
    print(str(error).splitlines()[0])
````

````python
try:
    getSubhaloStellarCirc('TNG100-1', 57, 92)
except KeyError as error:
    print(error)
````
